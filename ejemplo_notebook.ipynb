# Introducción a las Funciones en Python
Este cuaderno demuestra cómo definir y ejecutar funciones de prueba.

```python
def procesar_lista(numeros):
    # Multiplica cada número por 2
    return [x * 2 for x in numeros]

datos = [1, 2, 3, 4, 5]
resultado = procesar_lista(datos)
print("Resultado procesado:", resultado)
```
